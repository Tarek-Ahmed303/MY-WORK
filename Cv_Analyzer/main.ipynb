```python
import json

from document_parser import extract_from_file
from text_cleaner import clean_text
from section_detector import split_sections
from regex_extractor import extract_contact_information
from nlp_parser import extract_entities
from json_builder import build_resume


# ============================================================
# CV FILE
# ============================================================

file_path = r"E:\CVs\Mahmoud_Ahmed_Tarek_CV.docx"


# ============================================================
# 1. EXTRACT DOCUMENT TEXT
# ============================================================

raw_text = extract_from_file(file_path)


# ============================================================
# 2. CLEAN TEXT
# ============================================================

text = clean_text(raw_text)


# ============================================================
# 3. DETECT CV SECTIONS
# ============================================================

sections = split_sections(text)


# ============================================================
# 4. EXTRACT CONTACT INFORMATION
# ============================================================

contact = extract_contact_information(text)


# ============================================================
# 5. RUN NLP
# ============================================================

entities = extract_entities(text)


# ============================================================
# 6. BUILD FINAL JSON
# ============================================================

resume = build_resume(
    resume_id="resume_001",
    sections=sections,
    contact=contact,
    entities=entities
)


# ============================================================
# 7. SAVE JSON FILE
# ============================================================

output_file = "resume_output.json"

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(
        resume,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 8. PRINT RESULT
# ============================================================

print(f"JSON saved successfully to: {output_file}")
```
