```python
# ============================================================
# INITIALIZE RAG PIPELINE
# ============================================================

from pathlib import Path
import sys

# Notebook:

RAG_DIR = Path.cwd()

if str(RAG_DIR) not in sys.path:
    sys.path.insert(0, str(RAG_DIR))


# ------------------------------------------------------------
# IMPORTS
# ------------------------------------------------------------

from retrieval.hybrid_retriever import HybridRetriever
from retrieval.reranker import CrossEncoderReranker
from vector_store.Metadata_filter import MetadataFilter

from context.context_selector import ContextSelector
from context.context_builder import ContextBuilder


# ------------------------------------------------------------
# PATH
# ------------------------------------------------------------

CHUNKS_FILE = (
    RAG_DIR
    / "embeddings"
    / "processed"
    / "chunks"
    / "chunks.json"
)


# ------------------------------------------------------------
# INITIALIZE
# ------------------------------------------------------------

print("Initializing Hybrid Retriever...")
hybrid_retriever = HybridRetriever()

print("Initializing Reranker...")
reranker = CrossEncoderReranker()

print("Initializing Metadata Filter...")
metadata_filter = MetadataFilter(
    chunks_file=str(CHUNKS_FILE)
)

print("Initializing Context Selector...")
context_selector = ContextSelector(
    top_k=3
)

print("Initializing Context Builder...")
context_builder = ContextBuilder()


print("\n" + "=" * 70)
print("RAG COMPONENTS READY")
print("=" * 70)
```

```python

from answer.answerability import AnswerabilityGate
from llm.response_generator import ResponseGenerator
from pipeline import RAGPipeline

# Initialize the components
answerability_gate = AnswerabilityGate()
response_generator = ResponseGenerator()

print("Answerability Gate initialized.")
print("Response Generator initialized.")
```

```python

import importlib
import pipeline as pipeline_module

importlib.reload(pipeline_module)

RAGPipeline = pipeline_module.RAGPipeline

rag_pipeline = RAGPipeline(
    retriever=hybrid_retriever,
    reranker=reranker,
    metadata_filter=metadata_filter,
    context_selector=context_selector,
    context_builder=context_builder,
    answerability_gate=answerability_gate,  # Instance, not class
    response_generator=response_generator,
)

print("RAG pipeline rebuilt successfully.")
print("Updated pipeline ready.")

```

```python
question = "How does oxygen concentration affect flame behavior in microgravity?"

result = rag_pipeline.answer(question)

from IPython.display import display, JSON
display(JSON(result["response"]))
```
