```python
from pathlib import Path
import numpy as np
import faiss


# ============================================================
# Configuration
# ============================================================

EMBEDDINGS_FILE = Path(
    r"C:\Users\User\Nasa Rag SYS\rag\embeddings\processed\embeddings\embeddings.npy"
)

OUTPUT_DIR = Path(
    "processed/vector_store"
)

INDEX_FILE = OUTPUT_DIR / "faiss.index"


# ============================================================
# Load embeddings
# ============================================================

def load_embeddings():

    print("Loading embeddings...")

    embeddings = np.load(
        EMBEDDINGS_FILE
    )

    print(
        f"Loaded embeddings: {embeddings.shape}"
    )

    return embeddings


# ============================================================
# Build FAISS index
# ============================================================

def build_index(embeddings):

    dimension = embeddings.shape[1]

    print("\nBuilding FAISS index...")
    print(f"Dimension: {dimension}")

    # Inner Product + normalized vectors
    # = cosine similarity
    index = faiss.IndexFlatIP(
        dimension
    )

    index.add(
        embeddings.astype(np.float32)
    )

    print(
        f"Vectors added: {index.ntotal}"
    )

    return index


# ============================================================
# Save index
# ============================================================

def save_index(index):

    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    faiss.write_index(
        index,
        str(INDEX_FILE)
    )

    print(
        f"\nFAISS index saved to:\n"
        f"{INDEX_FILE.resolve()}"
    )


# ============================================================
# Validate index
# ============================================================

def validate_index(index, embeddings):

    print("\n================================")
    print("FAISS VALIDATION")
    print("================================")

    print(
        f"Index dimension: "
        f"{index.d}"
    )

    print(
        f"Number of vectors: "
        f"{index.ntotal}"
    )

    print(
        f"Expected vectors: "
        f"{len(embeddings)}"
    )

    if index.ntotal != len(embeddings):

        raise ValueError(
            "FAISS vector count does not "
            "match embedding count."
        )

    if index.d != embeddings.shape[1]:

        raise ValueError(
            "FAISS dimension does not "
            "match embedding dimension."
        )

    print("Vector count validation: PASSED")
    print("Dimension validation: PASSED")


# ============================================================
# Main
# ============================================================

if __name__ == "__main__":

    print("================================")
    print("FLAME IN FREEFALL FAISS")
    print("================================")

    # Load embeddings
    embeddings = load_embeddings()

    # Build index
    index = build_index(
        embeddings
    )

    # Validate
    validate_index(
        index,
        embeddings
    )

    # Save
    save_index(
        index
    )

    print("\n================================")
    print("FAISS BUILD COMPLETE")
    print("================================")
```
